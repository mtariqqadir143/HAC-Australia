```python
# lists, tuples, and dictionaries
# lists : collection of data elements that are defined inside []
# lists are mutable, which means you can change the elements inside the list
# work with multiple data types

list_x = [12,12.434,True,"London"]

print(list_x ,"and the data type is ", type(list_x))
```

```python
list_x = [12,12.434,True,"London"]

# accessing the elements
# index starts at 0

# extract the element at a index position : nameoflist[indexposition]

print("Element at index position 0 : ",list_x[0])

print("Element at index position 3 : ",list_x[3])
```

```python
# mutable : add,  remove, change , delete , update

list_x = [12,12.434,True,"London"]
print("Old list: ",list_x)

# add : .append() , only adds elements to the end of the list
# .append(), we can only add 1 item to the list at a moment

list_x.append("Mumbai")
print("Updated list:" ,list_x)


```

```python
# if we want to append or add multiple elements
# we will be using .extend()

list_x = [12,12.434,True,"London"]
print("Old list: ",list_x)

new_list = ["New York","Tokyo",90,False, 8931.132]

list_x.extend(new_list)
print("Updated list:" ,list_x)
```
